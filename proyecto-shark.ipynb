```python
import pandas as pd

df = pd.read_csv("GSAF5(Sheet1-GSAF) (1).csv", encoding='latin1', low_memory=False)

df.columns = df.columns.str.strip()

columns_i_want= ['Date', 'Year', 'Type', 'Country', 'State', 'Location', 'Activity', 'Name', 'Sex', 'Age', 'Injuries', 'Fatal Y/N', 'Especies','Source']

df = df[df.columns.intersection(columns_i_want)]

df = df.dropna(how='all')
df.head()
```

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip().str.lower()
df['type'] = df['type'].str.replace('UNprovoked', 'Unprovoked', case=False)

df['type'] = df['type'].str.lower()
print(df['type'].unique())
df.head()
```

```python
df.head()
```

```python
df.duplicated().any()
```

```python
df = df.map(lambda s: s.lower() if type(s) == str else s)

values_to_replace = ['?', 'not stated', 'no stated', 'not reported', 'unknown']
df = df.replace(values_to_replace, 'Unknown')

df = df.fillna('unknown')

df.head()
```

```python
df['sex'] = df['sex'].replace({'m': 'Male', 'f': 'Female'})

df.head()
```

```python
# Borrar una sola columna
df = df.drop('gender_mapped', axis=1)
df.head()

```
