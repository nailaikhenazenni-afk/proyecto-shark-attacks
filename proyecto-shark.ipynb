```python
import pandas as pd

df = pd.read_csv("GSAF5(Sheet1-GSAF) (1).csv", encoding='latin1', low_memory=False)

df.columns = df.columns.str.strip()

columns_i_want= ['Date', 'Year', 'Type', 'Country', 'State', 'Location', 'Activity', 'Age', 'Injury', 'Fatal Y/N', 'Especies','Source']

df = df[df.columns.intersection(columns_i_want)]

df = df.dropna(how='all')
df.head()
```

```python
df.columns = df.columns.str.strip().str.lower()
df['type'] = df['type'].str.replace('UNprovoked', 'Unprovoked', case=False)

df['type'] = df['type'].str.lower()
print(df['type'].unique())
df.head()
```

```python

df.columns = df.columns.str.lower().str.strip()

if 'state' in df.columns:
    unique_states = df['state'].unique()
    print(unique_states)
```

```python
df['state'] = df['state'].str.replace('NSW', 'New South Wales', case=False)

df.head()
```

```python
df['state'] = df['state'].replace('Between New Ireland & New Britain', 'New Ireland / Britain', case=False)

df['state'] = df['state'].str.title()
```

```python
values_to_replace = ['NaN']
df = df.replace(values_to_replace, 'Unknown')

df = df.fillna('Unknown')

df.head()
```

```python
unique_injuries= df['injury'].unique()

inspection_list = pd.Series(unique_injuries).dropna().sort_values()

print(inspection_list.to_string())
```

```python
df['injury'] = df['injury'].astype(str).str.strip()

df['injury'] = df['injury'].str.replace(r'\bR\b', 'right', case=False, regex=True)

df['injury'] = df['injury'].str.replace(r'\bL\b', 'left', case=False, regex=True)

df['injury'] = df['injury'].str.lower().str.capitalize()

print(df['injury'].unique()[:10]) # Vemos los primeros 10 para comparar
```
